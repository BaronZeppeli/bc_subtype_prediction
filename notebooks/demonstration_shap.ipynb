```python
import matplotlib.pyplot as plt
import explainability.shap_radiomics as shap_radiomics
```

```python
PLT_PARAMS = {
    "font.family": "Arial",
    "font.size": 12,
    "axes.titlesize": 16,
    "axes.labelsize": 16,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "legend.fontsize": 11,
    "lines.linewidth": 2
}
DPI = 100
plt.rcParams.update(PLT_PARAMS)

class_names = ["Luminal A", "Luminal B"]
max_display = 11
clustering_cutoff = 0.5
example_idx = 0
show = True
tvt = "train"

file_task = "../data/006_LumA_LumB_BBox_liminal_tvt_prior.json"

file_model_rf = "../models/random_forest/random_forest_model.pkl"
file_scaler_rf = "../models/random_forest/random_forest_scaler.pkl"
file_hyperparams_rf = "../models/random_forest/random_forest_hyperparams.json"
```

```python
# Single radiomics features.
%matplotlib
sr = shap_radiomics.SHAPRadiomics(
            file_task=file_task,
            model_id=None,
            tvt=tvt,
            summation_mode="single",
            file_weights=file_model_rf,
            file_scaler=file_scaler_rf,
            file_hyperparams=file_hyperparams_rf)
sr.visual_summary(max_display=max_display,
                  class_names=class_names,
                  example_idx=example_idx,
                  show=show)
```

```python
# Radiomics features summed up by modality.
%matplotlib
sr = shap_radiomics.SHAPRadiomics(
            file_task=file_task,
            model_id=None,
            tvt=tvt,
            summation_mode="sum",
            file_weights=file_model_rf,
            file_scaler=file_scaler_rf,
            file_hyperparams=file_hyperparams_rf)
sr.visual_summary(max_display=max_display,
                  class_names=class_names,
                  example_idx=example_idx,
                  show=show)
```

```python
# Mean contribution per modality.
%matplotlib
sr = shap_radiomics.SHAPRadiomics(
            file_task=file_task,
            model_id=None,
            tvt=tvt,
            summation_mode="mean",
            file_weights=file_model_rf,
            file_scaler=file_scaler_rf,
            file_hyperparams=file_hyperparams_rf)
sr.visual_summary(max_display=max_display,
                  class_names=class_names,
                  example_idx=example_idx,
                  show=show)
```
