```python
import torchvision.transforms as tf

import explainability.grad_cam as grad_cam
import explainability.center_periphery as center_periphery
import resnet_subtype.transform as transform
```

```python
resnet_models = {"dce_mri":
                     {"channel_couple": [0, 1, 2, 3, 4, 5, 6],
                      "file_weights": "../models/resnet/weights/dce_mri.pth",
                      "file_hyperparams": "../models/resnet/hyperparams/hyperparams_dce_mri.json"},
                 "t2w":
                     {"channel_couple": [7],
                      "file_weights": "../models/resnet/weights/t2w.pth",
                      "file_hyperparams": "../models/resnet/hyperparams/hyperparams_t2w.json"},
                 "adc":
                     {"channel_couple": [8],
                      "file_weights": "../models/resnet/weights/adc.pth",
                      "file_hyperparams": "../models/resnet/hyperparams/hyperparams_adc.json"},
                 "pet":
                     {"channel_couple": [9],
                      "file_weights": "../models/resnet/weights/pet.pth",
                      "file_hyperparams": "../models/resnet/hyperparams/hyperparams_pet.json"}
                 }
file_task = "../data/006_LumA_LumB_BBox_liminal_tvt_prior.json"
```

```python
VOXEL_SIZE_ORIGINAL = 1
PS = [0, 0.25, 0.5, 0.75]

file_task = "../data/006_LumA_LumB_BBox_liminal_tvt_prior.json"

for version in resnet_models.keys():
    tc = center_periphery.TumorCenter(file_task=file_task,
                                      model_id=version,
                                      voxel_size_original=VOXEL_SIZE_ORIGINAL)
    tc.get_center_periphery_shells(ps=PS)
```

```python
# Performing Grad-CAM.
target_size = 128
transform_2D = tf.Compose([
    transform.ResizeWithConstantRatio(target_size=target_size,
                                      interpolation_mode=None),
])
transform_3D = transform_2D

####

for version in resnet_models.keys():
    channel_couple = resnet_models[version]["channel_couple"]
    file_weights = resnet_models[version]["file_weights"]
    file_hyperparams = resnet_models[version]["file_hyperparams"]
    gcrn = grad_cam.GradCAMResNet(model_id=version,
                                  file_task=file_task,
                                  channel_couple=channel_couple,
                                  transform_2D=transform_2D,
                                  transform_3D=transform_3D,
                                  file_weights=file_weights,
                                  file_hyperparams=file_hyperparams)
    gcrn.export_cam()
    gcrn.export_mask_arrs()
```

```python
tvt = "train"
sample_id = "TUE018_001_0001"

egc = center_periphery.EvaluateGradCAM(file_task=file_task,
                                       dict_models=resnet_models)
egc.shell_plot_single(tvt=tvt,
                      sample_id=sample_id)
```
